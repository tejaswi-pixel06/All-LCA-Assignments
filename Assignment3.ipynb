```python
#Assignment 3
#Problem statement : To check whether or not the triangle is a right angled triangle using function

def is_right_triangle():
#Taking inputs from the user
    x=float(input("Enter the length of Base:"))
    y=float(input("Enter the length of Lenght:"))
    z=float(input("Enter the length of Hypotenues:"))

#Store them in a list and sort in ascending order
    sides = [x,y,z]
    sides.sort()

#Using pythagoras theorem
    if sides[0] ** 2 + sides[1] ** 2 == sides[2] ** 2:
        print("The Triangle is Right Angled.")
    else:
        print("The Triangle is not Right Angled.")

#Call the function
is_right_triangle()
```
